```python
import json
from agents.pipeline_coordinator import run_pipeline

# 1. Load data
jd = json.load(open("data/job_description.json"))
resumes = json.load(open("data/resumes.json"))["resumes"]
avail = json.load(open("data/availability.json"))

# 2. Run the 5-agent pipeline
print("Running pipeline...\n")
pool = run_pipeline(jd, resumes, avail, shortlist_size=3, enable_bias_reviewer=True)

# 3. Print the Multi-Agent Trace
print("=== MULTI-AGENT TRACE ===")
for line in pool.trace:
    print(line)

# 4. Print the Final Shortlist
print("\n=== RANKED SHORTLIST ===")
for c in sorted([c for c in pool.candidates if c.rank], key=lambda c: c.rank):
    if c.rank <= pool.shortlist_size:
        flag = f" [FLAGS: {c.bias_flags}]" if c.bias_flags else ""
        print(f"#{c.rank} {c.name} (score {c.match_score}) - {c.rank_rationale}{flag}")
        print(f"    Interview: {c.interview_slot or 'NO SLOT - escalate to human'}")
```
